# Task 1 — Show the twin-rotor model

# 1) Configuration and kinematics
From the PDF, the attitude is parameterized as
$$
R = R_1(\phi)\,R_3(\theta).
\$$


Differentiating gives:
$$
\dot R = R\,\hat\Omega,
\quad
\Omega =
\begin{bmatrix}
\dot\phi\cos\theta\\
-\dot\phi\sin\theta\\
\dot\theta
\end{bmatrix}.
$$


Define the *spatial* angular velocity
$$
\omega := R\Omega.
$$
Then use the identity
$$R\hat\Omega = \widehat{(R\Omega)}R$$ to rewrite kinematics as
$$
\boxed{\dot R = \hat\omega\,R.}
$$

# 2 Momentum relation and $\omega=(\mathbb I^R)^{-1}\pi$

$$
\pi = R\,\mathbb I\,\Omega.
$$


Since $\Omega = R^T\omega$,

$$
\pi = R\,\mathbb{I}\,\Omega
     = R\,\mathbb{I}\,(R^T\omega)
     = R\,\mathbb{I}\,R^T\,\omega.
$$

Define the *rotated (spatial) inertia*:

$$
\mathbb{I}^R := R\,\mathbb{I}\,R^T.
$$

So,

$$
\boxed{\pi = \mathbb{I}^R\omega \ \Rightarrow\  \omega = (\mathbb{I}^R)^{-1}\pi.}
$$


## 3) Dynamics split: $\dot{\pi}=\tau^e+\tau^u$

# (a) Constraint moment (unmanipulatable)

The constraint prevents rotation about $b_2$, so the torque has the form

$$
\boxed{\tau^e = R\begin{bmatrix}0\\T_2\\0\end{bmatrix}.}
$$

The PDF gives

$$
\boxed{T_2 = e_2^T\Big(\Omega\times(\mathbb{I}\Omega)+\mathbb{I}\dot{\Omega}\Big),}
\qquad \text{with}\qquad \Omega = R^T\omega.
$$

# (b) Actuation moment (manipulatable / control)

The implementable control torque is

$$
\tau^u = R
\begin{bmatrix}
1&0\\
0&0\\
0&1
\end{bmatrix}
\begin{bmatrix}
\cos\alpha & -\cos\beta\\
\sin\alpha & -\sin\beta
\end{bmatrix}
\begin{bmatrix}
u_1\\u_2
\end{bmatrix}.
$$

Equivalently,

$$
\boxed{
\tau^u = R
\begin{bmatrix}
u_1\cos\alpha-u_2\cos\beta\\
0\\
u_1\sin\alpha-u_2\sin\beta
\end{bmatrix}.
}
$$

Finally,

$$
\boxed{\dot{\pi} = \tau^u + \tau^e.}
$$

# Therefore

$$
\boxed{
\dot{R}=\hat{\omega}R,\qquad
\dot{\pi}=\tau^e+\tau^u,\qquad
\omega=(\mathbb{I}^R)^{-1}\pi
}
$$

<h2 style="text-align: center;"><u>Twin Rotor Model</u></h2>

## Task 1 — Show the twin-rotor model

## 1) Configuration and kinematics
From the PDF, the attitude is parameterized as

$$
R = R_1(\phi)\,R_3(\theta).
$$

Differentiating gives (PDF derivation):

$$
\dot{R} = R\,\hat{\Omega},
\qquad
\Omega =
\begin{bmatrix}
\dot{\phi}\cos\theta\\
-\dot{\phi}\sin\theta\\
\dot{\theta}
\end{bmatrix}.
$$

Define the *spatial* angular velocity

$$
\omega := R\Omega.
$$

Then use the identity $R\hat{\Omega}=\widehat{(R\Omega)}\,R$ to rewrite kinematics as

$$
\boxed{\dot{R} = \hat{\omega}\,R.}
$$

---

## 2) Momentum relation and $\omega=(\mathbb{I}^R)^{-1}\pi$
With constant body inertia $\mathbb{I}$ (in body frame), the spatial angular momentum is

$$
\pi = R\,\mathbb{I}\,\Omega.
$$

Since $\Omega = R^T\omega$,

$$
\pi = R\,\mathbb{I}\,R^T\,\omega.
$$

Define the *rotated (spatial) inertia*:

$$
\mathbb{I}^R := R\,\mathbb{I}\,R^T.
$$

So,

$$
\boxed{\pi=\mathbb{I}^R\omega \ \Rightarrow\  \omega=(\mathbb{I}^R)^{-1}\pi.}
$$

---

## 3) Dynamics split: $\dot{\pi}=\tau^e+\tau^u$

### (a) Constraint moment (unmanipulatable)
The constraint prevents rotation about $b_2$, so the torque has the form

$$
\boxed{\tau^e = R\begin{bmatrix}0\\T_2\\0\end{bmatrix}.}
$$

with

$$
\boxed{T_2 = e_2^T\Big(\Omega\times(\mathbb{I}\Omega)+\mathbb{I}\dot{\Omega}\Big),}
\qquad \text{and} \qquad \Omega = R^T\omega.
$$

---

### (b) Actuation moment (manipulatable / control)
The implementable control torque is

$$
\tau^u = R
\begin{bmatrix}
1&0\\
0&0\\
0&1
\end{bmatrix}
\begin{bmatrix}
\cos\alpha & -\cos\beta\\
\sin\alpha & -\sin\beta
\end{bmatrix}
\begin{bmatrix}
u_1\\u_2
\end{bmatrix}.
$$

Equivalently,

$$
\boxed{\tau^u = R
\begin{bmatrix}
u_1\cos\alpha-u_2\cos\beta\\
0\\
u_1\sin\alpha-u_2\sin\beta
\end{bmatrix}.}
$$

Finally, the equations of motion are

$$
\boxed{\dot{\pi} = \tau^u+\tau^e.}
$$

---

## Therefore, the twin-rotor system can be modeled by

$$
\boxed{
\dot{R}=\hat{\omega}R,\qquad
\dot{\pi}=\tau^e+\tau^u,\qquad
\omega=(\mathbb{I}^R)^{-1}\pi
}
$$

with $\tau^u,\tau^e,T_2$ as given above.
